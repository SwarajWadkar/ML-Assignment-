```python
import pandas as pd

# Load the dataset
df = pd.read_csv(
    "SMSSpamCollection",
    sep="\t",                    # tab separated
    names=["label", "message"]   # column names
)

df.head()
```

```python
df["label"] = df["label"].map({
    "spam": 1,
    "ham": 0
})

df.head()
```

```python
df.info()          # dataset info
df.isnull().sum()  # missing values
df["label"].value_counts()  # class balance
```

```python
X = df["message"]   # input text
y = df["label"]     # output class
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,        # 20% data for testing
    random_state=42,      # reproducibility
    stratify=y            # keeps spam/ham ratio same
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    stop_words="english",   # removes common words like 'the', 'is'
    max_features=5000       # limits vocabulary size
)
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
```

```python
X_test_tfidf = vectorizer.transform(X_test)
```

```python
from sklearn.svm import SVC

svm_model = SVC(
    kernel="linear",   # best for text
    C=1.0              # regularization strength
)

svm_model.fit(X_train_tfidf, y_train)
```

```python
y_pred = svm_model.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
new_emails = [
    "Congratulations! You have won a free iPhone",
    "Hey, are we meeting at 5pm today?",
    "Limited time offer, click now to claim prize"
]

# Convert text to TF-IDF
new_emails_tfidf = vectorizer.transform(new_emails)

# Predict
predictions = svm_model.predict(new_emails_tfidf)

predictions
```
