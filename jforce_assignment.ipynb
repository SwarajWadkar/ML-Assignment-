```python
import pandas as pd
```

```python
data = {
    "Roll No": [1, 1, 2, 2, 1, 2],
    "Subject": ["DS", "Java", "DS", "Java", "DS", "Java"],
    "Marks": [25, 40, 30, 26, 40, 50],
    "Date": ["01-01-23","02-01-23","01-01-23","02-01-23","15-01-23","15-01-23"]
    }
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df = df.sort_values(["Roll No", "Subject", "Date"],ascending=[True, True, False])
```

```python
print(df)
```

```python
df["Attempt"] = (df.groupby(["Roll No", "Subject"]).cumcount() + 1)
```

```python
print(df)
```

```python
df = df[df["Attempt"] <= 3]
```

```python
print(df)
```

```python
latest_date = (df[df["Attempt"] == 1].set_index(["Roll No", "Subject"])["Date"])
```

```python
print(latest_date)
```

```python
result = df.pivot_table(index=["Roll No", "Subject"],columns="Attempt",values="Marks",aggfunc="first")
```

```python
print(result)
```

```python
result.columns = [f"M{i}" for i in result.columns]

print(result)
```

```python
result = result.join(latest_date)

print(result)
```

```python
result = result.fillna(0)

print(result)
```

```python
for col in ["M1", "M2", "M3"]:
    if col not in result.columns:
        result[col] = 0
result["M1"] = result["M1"].astype(int)
result["M2"] = result["M2"].astype(int)
result["M3"] = result["M3"].astype(int)
```

```python
result = result.reset_index()
result = result[["Roll No", "Subject", "M1", "M2", "M3", "Date"]]
print(result)
```
